```python
# 計算每個品項的銷售數量並找出最受歡迎的品項
for product_name, group_data in df.groupby('Product'):
    total_sales = group_data['Quantity'].sum()
    # 使用 .iloc[0] 來安全地取得每個分組中 'Price' 欄位的第一個值
    price_per_item = group_data['Price'].iloc[0]
    print(f"產品: {product_name}, 總銷售數量: {total_sales} x 單價: {price_per_item} = {total_sales * price_per_item}")
```

## 繪圖

### 1. 繪製每個產品的總銷售數量長條圖

首先，我們來看看每個產品的總銷售數量分佈。

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 設定圖表風格
sns.set_style("whitegrid")

# 計算每個產品的總銷售數量 (若尚未計算，則在此處重新計算)
product_sales = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)

# 繪製長條圖
plt.figure(figsize=(10, 6))
sns.barplot(x=product_sales.index, y=product_sales.values, palette='viridis')
plt.title('Amount', fontsize=16)
plt.xlabel('Product Name', fontsize=12)
plt.ylabel('Total # saled item', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('product_sales.png') # before show fig
plt.show()
```

### 2. 繪製每個類別的銷售額圓餅圖

接下來，我們來看看不同產品類別的銷售額佔比。

```python
# 計算每個交易的總銷售額
df['TotalSales'] = df['Price'] * df['Quantity']

# 計算每個類別的總銷售額
category_sales = df.groupby('Category')['TotalSales'].sum().sort_values(ascending=False)

# 繪製圓餅圖
plt.figure(figsize=(8, 8))
plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.2f%%',
)
plt.title('Sales Mix', fontsize=16)
plt.ylabel('') # 移除Y軸標籤，因為圓餅圖不需要
plt.tight_layout()
plt.savefig('category_sales_pie.png') # 儲存圖片
plt.show()
```

### 3. 比較不同會員類別的總銷售額差異

現在，讓我們來看看會員和非會員的總銷售額有何不同。

```python
# 計算每個會員類別的總銷售額
member_sales = df.groupby('Member')['TotalSales'].sum().sort_values(ascending=False)

# 繪製圓餅圖
plt.figure(figsize=(8, 8))
plt.pie(
    member_sales.values,
    labels=member_sales.index,
    autopct='%1.2f%%',
)
plt.title('Total Sales by Member Type', fontsize=16)
plt.ylabel('') # 移除Y軸標籤，因為圓餅圖不需要
plt.tight_layout()
plt.savefig('member_sales_pie.png') # 儲存圖片
plt.show()
```

要賺大錢 == 會員多一點?

### 4. 分析會員與非會員的商品種類偏好

現在，我們將進一步分析會員和非會員在不同商品類別上的消費偏好。

```python
# 計算每個會員類別中，各商品類別的總銷售額
member_category_sales = df.groupby(['Member', 'Category'])['TotalSales'].sum().unstack(fill_value=0)

# 繪製堆疊長條圖或分組長條圖
# 將 DataFrame 轉換為適合繪圖的格式
member_category_sales_melted = member_category_sales.reset_index().melt('Member', var_name='Category', value_name='TotalSales')

plt.figure(figsize=(12, 7))
sns.barplot(
    x='Category',
    y='TotalSales',
    hue='Member',
    data=member_category_sales_melted,
    palette='viridis'
)
plt.title('Total Sales by Category and Member Type', fontsize=16)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Total Sales Amount', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Member')
plt.tight_layout()
plt.savefig('member_category_preference.png') # 儲存圖片
plt.show()
```
