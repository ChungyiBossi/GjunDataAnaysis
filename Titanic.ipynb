#### 甚麼是`fillna()`?

`fillna` 是 Pandas DataFrame 和 Series 物件的一個方法，用於處理資料中的缺失值 (missing values)，也就是 `NaN` (Not a Number) 值。它的主要功能是將這些缺失值替換為指定的值。

**為什麼要處理缺失值？**

如果資料中存在缺失值，模型訓練可能會失敗或產生不準確的結果。

**`fillna` 的主要用法：**

`fillna` 方法可以接受多種參數來指定替換缺失值的方式：

1.  **替換為單一常數值：**
    ```python
    df['column_name'].fillna(value, inplace=True)
    ```
    *   `value`: 你想要替換缺失值的常數 (例如 0, 平均值, 中位數, 眾數等)。
    *   `inplace=True`: 這個參數會直接修改原始 DataFrame，而不是返回一個新的 DataFrame。如果為 `False` (預設值)，則會返回一個新的 DataFrame。

    **範例：**
    *   `df['Age'].fillna(df['Age'].mean(), inplace=True)`: 將 'Age' 欄位中的缺失值替換為該欄位的平均值。
    *   `df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)`: 將 'Embarked' 欄位中的缺失值替換為該欄位的眾數 (最常出現的值)。

2.  **使用前一個有效值填充 (forward fill)：**
    ```python
    df['column_name'].fillna(method='ffill', inplace=True)
    ```
    *   `method='ffill'` (或 `method='pad'`): 會使用前一個非缺失值來填充當前的缺失值。

3.  **使用後一個有效值填充 (backward fill)：**
    ```python
    df['column_name'].fillna(method='bfill', inplace=True)
    ```
    *   `method='bfill'` (或 `method='backfill'`): 會使用後一個非缺失值來填充當前的缺失值。

#### `groupby().transform()`

`groupby().transform()` 是 Pandas 中一個強大的功能，它允許你在對 DataFrame 進行分組操作後，將結果返回為一個與原始 DataFrame **具有相同索引和行數**的 Series 或 DataFrame。這對於以下情況特別有用：

*   **填補缺失值**：如你所做，根據組內的統計量（例如中位數、平均值）來填補缺失值。
*   **數據標準化或歸一化**：在每個組內獨立地進行標準化，同時保持原始數據的結構。
*   **計算組內排名**：為每個組內的元素計算其在該組中的排名。

**核心思想**：對每個分組應用一個函數，然後將這個函數的結果廣播回原始 DataFrame 的相應位置，而不是像 `agg()` 那樣返回一個簡化的聚合結果。

這就是為什麼 `df_ml.groupby(['sex', 'pclass'])['age'].transform('median')` 能直接生成一個與 `df_ml['age']` 完全對齊的 Series，以便無縫地用於 `fillna` 操作。

`transform(function)`


*   **字符串**: Pandas 內建的許多聚合函數都可以作為字符串傳入，例如 `'sum'`, `'mean'`, `'median'`, `'min'`, `'max'`, `'std'`, `'var'`, `'count'` 等。Pandas 會自動調用對應的方法。
    *   **範例**: `df.groupby('Group')['Value'].transform('mean')`
*   **可調用對象 (Callable)**: 任何接受 Series 或 DataFrame (取決於 `transform` 應用於 Series 還是整個 DataFrame) 並返回 Series 或 DataFrame 的 Python 函數，包括 `lambda` 函數。
    *   **範例**: `df.groupby('Group')['Value'].transform(lambda x: (x - x.min()) / (x.max() - x.min()))` (組內 Min-Max 歸一化)
    *   **範例**: `df.groupby('Group')['Value'].transform(np.median)` (這就是你在程式碼中使用的形式)

### 5-2.選擇訓練的特徵

什麼是 One-Hot Encoding
* 在機器學習中，模型通常只能處理數值型資料，無法直接理解像 'male'/'female' 或 'S'/'C'/'Q' 這樣的文字類別。如果我們直接將這些類別轉換為單一數字（例如：男性=0，女性=1），模型可能會誤解這些數字之間有大小或順序關係，但實際上並沒有。

* One-Hot Encoding 的目的是將類別型特徵轉換為一種向量，使得模型能夠正確理解這些類別而不會引入不必要的順序關係。盡量保持其離散的特性。

* 例如：
    1. sex 欄位有 'male', 'female' 兩個類別。
        * 'male' 會變成 [1, 0] （假設 'male' 在第二個位置）
        * 'female' 會變成 [0, 1] （假設 'female' 在第一個位置）
    2. embarked 欄位有 'S', 'C', 'Q' 三個類別。
        * 'S' 會變成 [1, 0, 0]
        * 'C' 會變成 [0, 1, 0]
        * 'Q' 會變成 [0, 0, 1]

* 每個類別都會變成一個新的欄位，該欄位的值只有 0 或 1，表示樣本是否屬於該類別。

````python
# --- 類別特徵編碼與特徵選擇 ---

# 定義用於模型的特徵 (X) 和目標變數 (y)
# 'survived' 是目標變數
# 'pclass', 'sibsp', 'parch', 'fare', 'age' 是數值特徵
# 'sex', 'embarked', 'class', 'who_extended' 是需要 One-Hot 編碼的類別特徵
# 'alone', 'adult_male', 'has_deck' 是布林/二元特徵

features = [
    # 'pclass',
    'age', 'sibsp', 'parch', 'fare',
    'sex', 'embarked', 'class', 'who_extended',
    'alone',
    # 'adult_male', 'has_deck'
]

# 選取特徵和目標變數
X = df_ml[features]
y = df_ml['survived'] # ground truth
````

````python
# 識別類別型欄位
categorical_features = ['sex', 'embarked', 'class', 'who_extended']

# 使用 OneHotEncoder 進行編碼
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
X_encoded = encoder.fit_transform(X[categorical_features]) # 針對"類別行"處理

# 獲取新的特徵名稱
encoded_feature_names = encoder.get_feature_names_out(categorical_features) # 把每一行名稱列出來

# 將編碼後的特徵轉換為 DataFrame
X_encoded_df = pd.DataFrame(X_encoded, columns=encoded_feature_names, index=X.index)
display(X_encoded_df.head())
````

````python

# 將數值特徵和編碼後的類別特徵合併
X_numerical = X.drop(columns=categorical_features) # 丟掉 'sex', 'embarked', 'class', 'who_extended'
X_processed = pd.concat([X_numerical, X_encoded_df], axis=1) # 把剩下的欄位 + 新的encoded欄位

print("\n資料預處理完成。現在的特徵集 (X_processed) 的前5行：")
display(X_processed.head())
print(f"\n處理後的特徵維度: {X_processed.shape}")
````

````python
# 將資料切分為訓練集和測試集 (80% 訓練, 20% 測試)
X_train, X_test, y_train, y_test = train_test_split(
    X_processed, y, test_size=0.2, random_state=42, stratify=y
)

print(f"訓練集特徵形狀: {X_train.shape}")
print(f"測試集特徵形狀: {X_test.shape}")
print(f"訓練集目標形狀: {y_train.shape}")
print(f"測試集目標形狀: {y_test.shape}")

print("\n資料已成功切分為訓練集和測試集。接下來可以訓練 Random Forest 模型了！")
````

### 什麼是正則化 (Regularization)？

正則化是一種在機器學習模型訓練過程中，為了**防止過度擬合 (Overfitting)** 而引入的技術。

正則化的核心思想是**懲罰模型的複雜度**。模型越複雜，就越容易過度擬合。

正則化透過在目標函數（模型會試圖最小化的目標函數）中添加一個**懲罰項**來實現這一點。這個懲罰項會使得模型在優化過程中不僅要最小化預測誤差，還要最小化模型的複雜度。

**主要的正則化類型：**

1.  **L1 正則化 (Lasso Regularization)**：
    *   **懲罰項**：權重絕對值的和 (Sum of the absolute values of the weights)。
    *   **效果**：L1 正則化傾向於將某些特徵的權重**壓縮為零**，這意味著它可以進行**特徵選擇**。那些對模型貢獻不大的特徵會被完全剔除，從而產生一個更稀疏（較少特徵）的模型。
    image.png


2.  **L2 正則化 (Ridge Regularization)**：
    *   **懲罰項**：權重平方和 (Sum of the squared values of the weights)。
    *   **效果**：L2 正則化會將所有特徵的權重**趨向於零，但不會完全為零**。它使得模型中的權重值更小，從而降低模型的複雜度，減少個別特徵的影響力，但會保留所有特徵。
    image.png

### 7-3. KNeighborsClassifier



**n_neighbors ( K ):**
1. K 太小（例如 1）：模型極度敏感，容易受到雜訊（特徵極端值）影響，導致過擬合（Overfitting）。
2. K 太大（例如 100）：模型會被大群體給淹沒，邊界變得模糊，導致欠擬合（Underfitting）。
3. 通常選擇奇數（如 3, 5, 7），以避免投票時出現平手。

**metric ( 距離度量方式 )**
1. 歐幾里得距離 (Euclidean Distance)：即直連的直線距離。最常用（預設）。
2. 曼哈頓距離 (Manhattan Distance)：像在城市街道中只能走格子狀的距離。
3. 閔可夫斯基距離 (Minkowski Distance)：前兩者的通用公式，透過參數 p 切換（p=1 為曼哈頓，p=2 為歐幾里得）。

**weights ( 投票權重 )**
1. 'uniform'：每個鄰居都算 1 票（預設）。
2. 'distance'：距離越近的鄰居，票數權重越高（1/距離）。這能有效解決遠處鄰居干擾的問題。


#### KNN 的缺點與解決方法

因為 KNN 的這種獨特邏輯，它有兩個在實務上必須特別注意的致命傷：

1. 極度依賴「資料特徵縮放（Feature Scaling）」：如果特徵 A（如年齡：0-100）和特徵 B（如年薪：0-1,000,000）一起計算距離，年薪的差距會完全主導距離結果，使年齡失效。解法：使用 KNN 前，必須先進行標準化（Standardization/MinMax Scaling）。

2. 計算代價高：每次預測新資料，都要和資料庫中的每一筆資料重新計算一次距離。當資料量有百萬筆、特徵有幾百個時，預測會變得非常慢。

````python
knn = KNeighborsClassifier(
    n_neighbors=10
)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)
print("KNN 模型準確度 (Accuracy):", accuracy_score(y_test, y_pred)*100)
print("\n分類報告 (Classification Report):\n", classification_report(y_test, y_pred))
````

## 8.網格搜尋最佳參數

### 8-1.定義模型和它們的參數網格

````python
from sklearn.model_selection import GridSearchCV



# Logistic Regression
lr_param_grid = {
    'C': [0.01, 0.1, 1, 10, 100],
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear'] # 'liblinear' 支援 'l1' 和 'l2'
}

# Decision Tree Classifier
dtc_param_grid = {
    'max_depth': [None, 5, 10, 15, 20],
    'min_samples_leaf': [1, 2, 4], # 葉節點包含的最小sample數
    'criterion': ['gini', 'entropy'] # 衡量節點的不純度指標
}

# Random Forest Classifier
rf_param_grid = {
    'n_estimators': [100, 200, 300, 500, 1000, 2000],
    'max_depth': [None, 10, 20],
    'min_samples_leaf': [1, 2, 4]
}

# KNeighbors Classifier
knn_param_grid = {
    'n_neighbors': [3, 5, 7, 9, 11],
    'weights': ['uniform', 'distance'],
    'p': [1, 2] # p=1 for Manhattan distance, p=2 for Euclidean distance
}

models = {
    'Logistic Regression': (LogisticRegression(random_state=42), lr_param_grid),
    'Decision Tree': (DecisionTreeClassifier(random_state=42), dtc_param_grid),
    'Random Forest': (RandomForestClassifier(random_state=42), rf_param_grid),
    'KNeighbors': (KNeighborsClassifier(), knn_param_grid)
}

print("開始進行網格搜尋以尋找最佳參數...")

best_models = {}
for model_name, (model, param_grid) in models.items():
    print(f"\n正在為 {model_name} 進行網格搜尋...")
    grid_search = GridSearchCV(
        model,
        param_grid,
        cv=5, # 交叉驗證
        scoring='accuracy',
        n_jobs=-1,
        verbose=1
    )
    grid_search.fit(X_train, y_train)

    best_models[model_name] = grid_search

    print(f"最佳的 {model_name} 參數: {grid_search.best_params_}")
    print(f"最佳的 {model_name} 準確度: {grid_search.best_score_:.4f}")

print("\n所有模型的網格搜尋已完成！")

# 您可以透過 best_models['Model Name'].best_estimator_ 來獲取最佳模型
````

````python
# 找出最佳閾值 (這裡以 F1-score 最大化為例)
optimal_threshold_idx = np.argmax(f1_scores) # 找到最大值的index
optimal_threshold = thresholds[optimal_threshold_idx]
optimal_f1_score = f1_scores[optimal_threshold_idx]

# 在最佳閾值下的預測
y_pred_optimal = (y_pred_proba >= optimal_threshold).astype(int)
optimal_accuracy = accuracy_score(y_test, y_pred_optimal)

print(f"根據 F1-score 最大化，最佳閾值為: {optimal_threshold:.4f}")
print(f"此最佳閾值下的 F1-score 為: {optimal_f1_score:.4f}")
print(f"此最佳閾值下的準確度為: {optimal_accuracy:.4f}")

````

````python
# 繪製 F1-score 隨閾值變化的圖表
plt.figure(figsize=(10, 6))
plt.plot(thresholds, f1_scores, label='F1-score')
plt.plot(thresholds, precisions[:-1], label='Precision')
plt.plot(thresholds, recalls[:-1], label='Recall')

# 畫直線
plt.axvline(
    x=optimal_threshold,
    color='r',
    linestyle='--',
    label=f'Optimal Threshold ({optimal_threshold:.2f})'
)
plt.title('F1-score, Precision and Recall vs. Threshold')
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.legend()
plt.grid(True)
plt.show()
````

#### 8-3-4. ROC 曲線


ROC 曲線是一種圖形，用於展示分類模型在所有分類閾值下的性能。它描繪了兩個指標之間的關係：

1.  **真陽性率 (True Positive Rate, TPR)**，也稱為**召回率 (Recall)** 或 **敏感度 (Sensitivity)**：
    *   `TPR (Recall) = TP / (TP + FN)`
2.  **假陽性率 (False Positive Rate, FPR)**，也稱為 **1 - 特異度 (1 - Specificity)**：
    *   `FPR = FN / (TP + FN)`
**如何繪製 ROC 曲線：**

*   模型會對每個樣本輸出一個預測概率。
*   我們遍歷所有可能的分類閾值（從 0 到 1）。
*   對於每個閾值，計算出對應的 TPR 和 FPR。
*   將 (FPR, TPR) 點繪製在二維座標系上，就得到了 ROC 曲線。

**理想的 ROC 曲線：**

*   一個完美的分類器，其 ROC 曲線會經過 (0, 1) 點，表示 TPR = 1 且 FPR = 0（即在沒有任何誤報的情況下，所有正樣本都被正確識別）。
*   一條對角線（從 (0, 0) 到 (1, 1)）表示一個隨機分類器，其性能與隨機猜測無異。
*   模型曲線越靠近左上角，表示模型性能越好。


#### 8-3-5. AUC (Area Under ROC Curve)

AUC 是 ROC 曲線下方的面積。它提供了一個單一的數值來總結模型在所有可能閾值下的整體性能。

*   **AUC 值範圍**：通常介於 0.5 和 1.0 之間。
    *   **AUC = 1.0**：表示模型是完美的，能夠完全區分正負樣本。
    *   **AUC = 0.5**：表示模型性能等同於隨機猜測。
    *   **AUC < 0.5**：表示模型性能比隨機猜測還差，可能是模型學到了錯誤的模式，或者分類標籤被反轉了。

**AUC 的解釋：**

AUC 可以被解釋為：隨機選擇一個正樣本和一個負樣本，模型將正樣本的預測概率（或分數）判斷高於負樣本的概率。AUC 值越高，模型區分正負樣本的能力越強。

**優勢：**

*   **對類別不平衡不敏感**：與準確度不同，AUC 不受數據集中正負樣本比例的影響。即使正樣本非常稀少，AUC 仍然能有效評估模型性能。
*   **閾值獨立**：AUC 衡量的是模型在所有可能閾值下的整體性能，而不是單一特定閾值下的性能。這使得它成為一個更全面的性能指標。

總之，ROC 曲線提供了模型在不同閾值下的視覺化性能，而 AUC 則將這種性能量化為一個單一的數值，方便比較不同模型。

````python
from sklearn.metrics import roc_curve, auc

# 獲取最佳的 Logistic Regression 模型
best_lr_model = best_models['Logistic Regression'].best_estimator_

# 獲取模型在測試集上的預測概率（針對正類別）
y_pred_proba_lr = best_lr_model.predict_proba(X_test)[:, 1]

# 計算 ROC 曲線的 FPR 和 TPR
fpr_lr, tpr_lr, thresholds_lr = roc_curve(y_test, y_pred_proba_lr)

# 計算 AUC (Area Under the Curve)
roc_auc_lr = auc(fpr_lr, tpr_lr)
````

````python
# 繪製 ROC 曲線
plt.figure(figsize=(8, 6))
plt.plot(fpr_lr, tpr_lr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc_lr:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random Classifier (area = 0.50)')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR)')
plt.title('Receiver Operating Characteristic (ROC) Curve for Logistic Regression')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

print(f"Logistic Regression 模型的 AUC 值: {roc_auc_lr:.4f}")
````
